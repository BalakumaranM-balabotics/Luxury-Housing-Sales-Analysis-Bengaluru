```python
clean_df = lhb_df.copy()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop_duplicates()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df["Property_ID"].duplicated().sum()
```

```python
clean_df["Micro_Market"].value_counts()
```

```python
#Changing all the values in micro market to upper case for consistency
clean_df["Micro_Market"] = clean_df["Micro_Market"].str.upper()
```

```python
#Total values for each micro market
clean_df["Micro_Market"].value_counts()
```

```python
#Changing Configuration values to upper case 
clean_df["Configuration"] = clean_df["Configuration"].str.upper()
```

```python
#Total values for each configuration
clean_df["Configuration"].value_counts()
```

```python
#clean_df.to_csv("Luxury_Housing_Bangalore_cleaning_checkpoint.csv", index=False)
```

```python
clean_df["Ticket_Price_Cr"].dtype
```

```python
#removing ₹ 
clean_df["Ticket_Price_Cr"] = clean_df["Ticket_Price_Cr"].str.replace("₹", "")
```

```python
#removing "Cr"
clean_df["Ticket_Price_Cr"] = clean_df["Ticket_Price_Cr"].str.replace(" Cr", "")
```

```python
#converting to float
clean_df["Ticket_Price_Cr"] = pd.to_numeric(clean_df["Ticket_Price_Cr"], errors="coerce")
```

```python
#Negative values in Ticket price
clean_df[clean_df["Ticket_Price_Cr"] < 0]
```

```python
#converting negative values to NaN in ticket price column
import numpy as np
clean_df.loc[clean_df["Ticket_Price_Cr"] < 0, "Ticket_Price_Cr"] = np.nan
```

```python
#checking the null values in ticket price column after converting negative values to NaN
clean_df["Ticket_Price_Cr"].isnull().sum()
```

```python
#first 5 rows of the null values in ticket price column after converting negative values to NaN
clean_df[clean_df["Ticket_Price_Cr"].isnull()].head()
```

```python
#negative values in unit size column
(clean_df["Unit_Size_Sqft"] < 0).sum()
```

```python
#Negative value and how many of them
clean_df[clean_df["Unit_Size_Sqft"] < 0]["Unit_Size_Sqft"].value_counts()
```

```python
#converting negative values to NaN in unit size column
import numpy as np
clean_df.loc[clean_df["Unit_Size_Sqft"] <= 0, "Unit_Size_Sqft"] = np.nan
```

```python
#total null values in unit size column after converting negative values to NaN
clean_df["Unit_Size_Sqft"].isnull().sum()
```

```python
#median unit size for each configuration
clean_df.groupby("Configuration")["Unit_Size_Sqft"].median()
```

```python
#Filling median values for null values in unit size column based on configuration
clean_df["Unit_Size_Sqft"] = clean_df.groupby("Configuration")["Unit_Size_Sqft"].transform(lambda x: x.fillna(x.median()))
```

```python
#checking null values in unit size 
clean_df["Unit_Size_Sqft"].isnull().sum()
```

```python
#checking null values in Amenity score column
clean_df["Amenity_Score"].isnull().sum()
```

```python
#checking the median value of Amenity score
clean_df["Amenity_Score"].median()
```

```python
#Filling null values with median value in Amenity score column
clean_df["Amenity_Score"] = clean_df["Amenity_Score"].fillna(clean_df["Amenity_Score"].median())
```

```python
#checking null values in Amenity score column
clean_df["Amenity_Score"].isnull().sum()
```

```python
#checking null values in Buyer comments column
clean_df["Buyer_Comments"].isnull().sum()
```

```python
#Filling empty values in Buyer comments column with "data_not_given"
clean_df["Buyer_Comments"] = clean_df["Buyer_Comments"].fillna("data_not_given")
```

```python
#checking null values in Buyer comments column
clean_df["Buyer_Comments"].isnull().sum()
```

```python
#checking null values in all the coloumns
clean_df.isnull().sum()
```
